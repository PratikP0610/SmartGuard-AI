```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(8,6))

correlation = data.corr()

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm"
)

plt.title("Feature Correlation Heatmap")

plt.show()
```

```python
plt.figure(figsize=(8,5))

data.boxplot(
    column="Vibration",
    by="Machine_Failure"
)

plt.title("Vibration Distribution vs Machine Failure")
plt.suptitle("")
plt.xlabel("Machine Failure (0 = Healthy, 1 = Failed)")
plt.ylabel("Vibration Level")

plt.show()
```
